# Data Description

The dataset used for this visualization assignment contains *automobile_sales* data representing automobile sales and related variables during recession and non-recession period. 

The dataset includes the following variables:
<br>1. Date: The month end date of sales observation.
<br>2. Recession: A binary variable indicating recession perion; 1 means it was recession, 0 means it was normal.
<br>3. Automobile_Sales: The number of vehicles sold during the period.
<br>4. GDP: The per capita GDP value in USD.
<br>5. Unemployment_Rate: The monthly unemployment rate.
<br>6. Consumer_Confidence: A synthetic index representing consumer confidence, which can impact consumer spending and automobile purchases.
<br>7. Seasonality_Weight: The weight representing the seasonality effect on automobile sales during the period. This variable represents the seasonal effect on automobile sales for a given month. In the automobile industry, sales often fluctuate throughout the year due to seasonal patterns—for example, sales may increase during the festive season or year-end promotions, and decrease during off-peak months like post-holiday winter periods.
A value greater than 1 indicates higher-than-average sales expected for that month due to seasonal trends (e.g., holiday season or new model launches).
A value less than 1 suggests lower-than-average sales due to seasonal slowdowns.
A value around 1 means the season has neutral or average effect on sales.
<br>8. Price: The average vehicle price during the period.
<br>9. Advertising_Expenditure: The advertising expenditure of the company.
<br>10.Vehicle_Type: The type of vehicles sold; Supperminicar, Smallfamiliycar,                 Mediumfamilycar, Executivecar, Sports.
<br>11.Competition: The measure of competition in the market, such as the number of competitors or market share of major manufacturers.
<br>12.Month: Month of the observation extracted from Date..
<br>13.Year: Year of the observation extracted from Date.
<br>
By examining various factors mentioned above from the dataset, you aim to gain insights into how recessions impacted automobile sales for your company.

####  TASK 1.2: How do trends in advertising expenditure correlate with automobile sales during non-recession periods, and what insights can be derived from this relationship.


<details><summary>Click here for a hint</summary>

<p>
    You will require:-
<br>Start by filtering the dataset to include only non-recession periods where the "Recession" column has a value of 0.
<br>Group the data by Year and calculate the average of "Automobile_Sales" and "Advertising_Expenditure".
<br>Ensure you reset the index after aggregation so the DataFrame remains structured properly.
<br>Use sns.lineplot() to plot trends:
<br>One line for average automobile sales (solid green line with circular markers).
<br>Another line for average advertising expenditure (dashed blue line with square markers).
<br>Add labels and a title to make the visualization clear.
<br>Include a legend and a grid for better readability.
</p>
</details>


````python
import matplotlib.pyplot as plt

# 1. Filter the data for non-recession periods (Recession == 0)
non_recession_df = df[df['Recession'] == 0]

# 2. Group by Year to get the average Sales and Advertising Expenditure per year
yearly_trends = non_recession_df.groupby('Year')[['Automobile_Sales', 'Advertising_Expenditure']].mean()

# 3. Plot the data
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot Automobile Sales on the primary y-axis
color = 'navy'
ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('Average Automobile Sales', color=color, fontsize=12)
line1 = ax1.plot(yearly_trends.index, yearly_trends['Automobile_Sales'], color=color, marker='o', label='Automobile Sales')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, linestyle='--', alpha=0.5)

# Create a secondary y-axis to plot Advertising Expenditure
ax2 = ax1.twinx()  
color = 'darkorange'
ax2.set_ylabel('Average Advertising Expenditure (USD)', color=color, fontsize=12)
line2 = ax2.plot(yearly_trends.index, yearly_trends['Advertising_Expenditure'], color=color, marker='s', linestyle='--', label='Advertising Expenditure')
ax2.tick_params(axis='y', labelcolor=color)

# Added a combined legend
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Advertising Expenditure vs. Automobile Sales During Non-Recession Periods', fontsize=14, fontweight='bold')
plt.show()
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Group the data by Recession and Vehicle_Type to get average sales
# This prepares the specific data points needed for the comparison
grouped_df = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

# 2. Set up the plotting environment
plt.figure(figsize=(10, 6))

# 3. Create a grouped bar chart using Seaborn
# x handles the vehicle categories, y tracks sales, and hue splits them by economic period
sns.barplot(
    data=grouped_df, 
    x='Vehicle_Type', 
    y='Automobile_Sales', 
    hue='Recession', 
    palette='Set2'
)

# 4. Refine titles, labels, and legends for presentation
plt.title('Average Automobile Sales Trend per Vehicle Type: Recession vs. Non-Recession', fontsize=14, fontweight='bold')
plt.xlabel('Vehicle Type', fontsize=12)
plt.ylabel('Average Automobile Sales', fontsize=12)

# Map the binary legend labels (0 and 1) to meaningful text
plt.legend(title='Economic Period', labels=['Non-Recession (0)', 'Recession (1)'])
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 5. Display the final plot
plt.show()
````

### TASK 1.4: Use sub plotting to compare the variations in GDP during recession and non-recession period by developing line plots for each period.
<br>Now, you want to find more insights from the data to understand the reason. <br>Plot a two line charts using subplotting to answer:-
#### How did the GDP vary over time during recession and non-recession periods? 
<br>Make use of <u>add_subplot()</u> from Matplotlib for this comparision.


````python
import matplotlib.pyplot as plt

# 1. Separate the data based on economic periods
recession_data = df[df['Recession'] == 1].groupby('Year')['GDP'].mean()
non_recession_data = df[df['Recession'] == 0].groupby('Year')['GDP'].mean()

# 2. Create the main figure container
fig = plt.figure(figsize=(14, 6))

# 3. Add the first subplot (Left: Recession Period GDP)
ax1 = fig.add_subplot(1, 2, 1)  # 1 row, 2 columns, position 1
ax1.plot(recession_data.index, recession_data.values, marker='o', color='darkred', linewidth=2)
ax1.set_title('GDP Variation During Recession Periods', fontsize=12, fontweight='bold')
ax1.set_xlabel('Year')
ax1.set_ylabel('Per Capita GDP (USD)')
ax1.grid(True, linestyle='--', alpha=0.5)

# 4. Add the second subplot (Right: Non-Recession Period GDP)
ax2 = fig.add_subplot(1, 2, 2)  # 1 row, 2 columns, position 2
ax2.plot(non_recession_data.index, non_recession_data.values, marker='s', color='darkgreen', linewidth=2)
ax2.set_title('GDP Variation During Non-Recession Periods', fontsize=12, fontweight='bold')
ax2.set_xlabel('Year')
ax2.set_ylabel('Per Capita GDP (USD)')
ax2.grid(True, linestyle='--', alpha=0.5)

# 5. Optimize layout and display
plt.tight_layout()
plt.show()
````

### TASK 1.5: Develop a Bubble plot for displaying the impact of seasonality on Automobile Sales.
<br>How has seasonality impacted the sales, in which months the sales were high or low? Check it for non-recession years to understand the trend

##### Develop a Bubble plot for displaying Automobile Sales for every month and use Seasonality Weight for representing the size of each bubble<br>
Title this plot as 'Seasonality impact on Automobile Sales'


<details><summary>Click here for a hint</summary>

<p>
     You can create Bubble Chart by calling the scatter()
        <br>Pass the 'Month' and 'Automobile_Sales' to the functions as x and y and then use Seasonality weight for size parameter</p>
</details>


````python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Filter for non-recession years
non_recession_df = df[df['Recession'] == 0]

# 2. Group by Month to find the average Sales and Seasonality Weight
monthly_seasonality = non_recession_df.groupby('Month')[['Automobile_Sales', 'Seasonality_Weight']].mean().reset_index()

# 3. Create the bubble plot
plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=monthly_seasonality,
    x='Month',
    y='Automobile_Sales',
    size='Seasonality_Weight',
    sizes=(100, 800),  # Sets the min and max size boundaries for the bubbles
    hue='Seasonality_Weight',
    palette='viridis',
    alpha=0.8
)

# 4. Refine the visual layout
plt.title('Seasonality impact on Automobile Sales', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Automobile Sales', fontsize=12)

# Move the legend outside so it doesn't overlap data points
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title='Seasonality Weight')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

 ### TASK 1.9: Develop a lineplot to analyse the effect of the unemployment rate on vehicle type and sales during the Recession Period.
 <br>Analyze the effect of the unemployment rate on vehicle type and sales during the Recession Period
 #### You can create a lineplot and title the plot as 'Effect of Unemployment Rate on Vehicle Type and Sales'


<details><summary>Click here for a hint</summary>

<p>
    Filter out the data for recession period<br>
     Make use of lineplot() from seaborn and pass the relavent data</p>
</details>


````python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Filter out the data for the recession period
recession_df = df[df['Recession'] == 1]

# 2. Use the correct lowercase column name 'unemployment_rate'
# Pre-aggregate data to get a clean trend line
aggregated_data = recession_df.groupby(['unemployment_rate', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

# 3. Set up the plotting environment
plt.figure(figsize=(12, 6))

# 4. Generate the lineplot
sns.lineplot(
    data=aggregated_data,
    x='unemployment_rate',
    y='Automobile_Sales',
    hue='Vehicle_Type',
    marker='o'
)

# 5. Finalize formatting
plt.title('Effect of Unemployment Rate on Vehicle Type and Sales', fontsize=14, fontweight='bold')
plt.xlabel('Unemployment Rate (%)', fontsize=12)
plt.ylabel('Average Automobile Sales', fontsize=12)
plt.legend(title='Vehicle Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

### OPTIONAL : TASK 1.10 Create a map on the hightest sales region/offices of the company during recession period


````python
from pyodide.http import pyfetch

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())

path = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/us-states.json'
await download(path, "us-states.json")

filename = "us-states.json"
````

#### You found that the datset also contains the location/city for company offices. Now you want to show the recession impact on various offices/city sales by developing a choropleth


````python
import json
import folium
import pandas as pd

# 1. Filter data for recession periods
recession_df = df[df['Recession'] == 1]

# 2. Calculate the total sales grouped by the correct column 'City'
sales_by_city = recession_df.groupby('City')['Automobile_Sales'].sum().reset_index()

# 3. Load the downloaded GeoJSON file
with open("us-states.json", "r") as f:
    us_geo = json.load(f)

# 4. Initialize a base map centered on the United States
map1 = folium.Map(location=[37.0902, -95.7129], zoom_start=4)

# 5. Build and add the Choropleth layer
folium.Choropleth(
    geo_data=us_geo,                  # GeoJSON file with state boundaries
    name='choropleth',
    data=sales_by_city,
    columns=['City', 'Automobile_Sales'],
    key_on='feature.properties.name', # Binds the dataframe 'City' to GeoJSON state names
    fill_color='YlOrRd',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Automobile Sales during Recession'
).add_to(map1)

# 6. Display the map
map1
````
